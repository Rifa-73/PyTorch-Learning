# Creating a Synthetic Dataset

We can use Scikit-learn's `make_classification()` function to generate
a small artificial classification dataset.

### Syntax

```python

    n_samples,
    n_features,
    n_informative,
    n_redundant,
    n_classes,
    random_state
)

| Parameter       | Meaning                                                                        |
| --------------- | ------------------------------------------------------------------------------ |
| `n_samples`     | Number of data points/samples                                                  |
| `n_features`    | Total number of input features                                                 |
| `n_informative` | Number of features that actually contain useful information for classification |
| `n_redundant`   | Number of redundant features created from informative features                 |

#  What is a Dataset?

A **Dataset** is an object that stores and provides access to our data.

In PyTorch, the Dataset abstraction allows us to:

- Store features and labels.
- Find the total number of samples.
- Retrieve an individual sample.
- Keep data handling separate from the model.

PyTorch provides:

```python
torch.utils.data.Dataset

````python
from torch.utils.data import Dataset, DataLoader
````

# Creating a Custom Dataset

We create our own Dataset class by inheriting from PyTorch's `Dataset`.

### Syntax

```python
class CustomDataset(Dataset):

````python
class CustomDataset(Dataset):

  def __init__(self, features, labels):

    self.features = features
    self.labels = labels

  def __len__(self):

    return self.features.shape[0]

  def __getitem__(self, index):

    return self.features[index], self.labels[index]
````

#  Dataset vs DataLoader

| Dataset | DataLoader |
|---|---|
| Stores/provides access to data | Loads data from Dataset |
| Works mainly with individual samples | Works mainly with batches |
| Implements `__len__()` | Uses Dataset length |
| Implements `__getitem__()` | Calls `__getitem__()` to obtain samples |
| Defines how data is accessed | Defines how data is batched/iterated |
| Example: `dataset[2]` | Example: `for batch in dataloader` |

### Easy way to remember

**Dataset = Where/how to get one sample**

**DataLoader = How to deliver samples to the model**





#  Important Definitions — Quick Revision

### Dataset

A PyTorch Dataset is an abstraction that provides access to individual
samples and their corresponding labels.

### Custom Dataset

A user-defined Dataset class created by inheriting from PyTorch's
`Dataset` class.

### `__init__()`

Initializes the Dataset and stores the required data.

### `__len__()`

Returns the total number of samples in the Dataset.

### `__getitem__()`

Returns one sample and its corresponding label using an index.

### DataLoader

A PyTorch utility that loads data from a Dataset and organizes it into
batches for efficient processing.

### Batch

A small group of samples processed together by the model.

### Batch Size

The number of samples contained in one batch.

### Shuffle

Determines whether the order of samples is randomized before batching.

#  Why Dataset and DataLoader Are Important

When training a neural network on a small dataset, we could technically
pass the entire dataset directly to the model.

However, real-world datasets can contain:

- Thousands of images
- Millions of text samples
- Large numerical datasets
- Large audio/video datasets

Loading everything at once may consume too much memory.

Dataset + DataLoader solve this problem by allowing us to:

- Load data in batches.
- Control batch size.
- Shuffle training data.
- Process large datasets efficiently.
- Separate data preparation from model architecture.

This becomes especially important when working with:

- CNNs
- RNNs
- Transformers
- Computer Vision datasets
- NLP datasets
- Large-scale Deep Learning projects
