# Building Neural Networks using nn.Module

In this session, I m learning how to create a neural network using PyTorch's `nn.Module`.

Here, I will use PyTorch's built-in neural network components such as:

- `nn.Module`
- `nn.Sequential`
- `nn.Linear`
- `nn.ReLU`
- `nn.Sigmoid`

This makes defining neural networks much easier and cleaner.

```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  def forward(self, features):

    out = self.network(features)

    return out
```

## Understanding the Model Architecture

The `Model` class inherits from `nn.Module`, which is the base class for neural networks in PyTorch.

The network contains the following layers:

1. `nn.Linear(num_features, 3)` — takes the input features and produces 3 outputs.
2. `nn.ReLU()` — applies the ReLU activation function.
3. `nn.Linear(3, 1)` — converts the 3 values into a single output.
4. `nn.Sigmoid()` — converts the output into a value between 0 and 1.

The overall architecture is:

Input Features → Linear → ReLU → Linear → Sigmoid → Output

```python
# create dataset
features = torch.rand(10,5)
```

## Creating a Sample Dataset

Here, we create a random tensor to test our neural network.

The tensor has a shape of `(10, 5)`:

- `10` → number of samples
- `5` → number of features for each sample

Therefore, each sample contains 5 input features.

```python
# create model
model = Model(features.shape[1])
```

## Creating the Model

We create an object of our `Model` class.

`features.shape[1]` gives the number of input features, which is `5` in this case.

Therefore, the first linear layer receives 5 input features.

```python
# call model for forward pass
# model.forward(features)
model(features)
```

## Forward Pass

The model can be called directly using:

`model(features)`

This internally calls the `forward()` method of the model.

The input data passes through all the layers defined inside `nn.Sequential`.

The data flows as:

Input → Linear → ReLU → Linear → Sigmoid → Prediction

```python
# show model weights
model.network[2].weight
```

## Inspecting Model Parameters

PyTorch automatically keeps track of the parameters of layers such as `nn.Linear`.

These parameters include:

- Weights
- Biases

We can inspect the weights of a layer using its corresponding attribute.

**Note:** In this particular implementation, the linear layers are stored inside `self.network` using `nn.Sequential`.
