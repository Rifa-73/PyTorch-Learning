# Breast Cancer Classification using PyTorch

In this notebook, we will build a basic binary classification model using PyTorch.

```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

## Dataset Shape

Let's check the number of rows and columns in the dataset.

```python
df.shape
```

## Data Cleaning

The `id` and `Unnamed: 32` columns are removed because they are not required for model training.

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace= True)

df.head()
```

## Train-Test Split

The dataset is divided into training and testing sets.

- 80% of the data is used for training.
- 20% of the data is used for testing.

```python
### train test split

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

## Feature Scaling

`StandardScaler` is used to standardize the input features.

The scaler is fitted on the training data and then used to transform both the training and testing data.

```python
### scaling

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

X_train
```

## Training Labels

Let's look at the training labels before applying label encoding.

```python
y_train
```

## Label Encoding

The categorical target labels are converted into numerical values using `LabelEncoder`.

For example:

- `B` → 0
- `M` → 1

```python
### Label Encoding

encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)

y_train
```

## Converting NumPy Arrays to PyTorch Tensors

PyTorch works with tensors, so the NumPy arrays are converted into PyTorch tensors.

```python
### Numpy arrays to PyTorch tensors

X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

## Checking Tensor Shapes

Let's check the shape of the training feature tensor.

```python
X_train_tensor.shape
```

Let's check the shape of the training target tensor.

```python
y_train_tensor.shape
```

## Defining the Model

We now define a simple neural network manually using PyTorch.

The model contains trainable weights and bias. The `forward()` method generates predictions, while the `loss_function()` calculates the training loss.

```python
### Defining the model

class MySimpleNN():

  def __init__(self, X):

    self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True)
    self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)

  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss
```

## Important Parameters

We define the learning rate and number of epochs for training.

```python
### Important Parameters

learning_rate = 0.1
epochs = 25
```

## Training Pipeline

The model is trained using a manual training loop.

Each epoch consists of:

1. Forward pass
2. Loss calculation
3. Backward pass
4. Parameter update
5. Gradient reset

```python
### Training Pipeline

# create model
model = MySimpleNN(X_train_tensor)

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model.forward(X_train_tensor)

  # loss calculate
  loss = model.loss_function(y_pred, y_train_tensor)

  # backward pass
  loss.backward()

  # parameters update
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

## Learned Bias

Let's inspect the bias value after training.

```python
model.bias
```

## Evaluation

Now we evaluate the trained model on the test dataset.

`torch.no_grad()` is used because gradients are not required during evaluation.

The model output is converted into binary predictions using a threshold of `0.9`.

```python
### Evaluation

# model evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.9).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
